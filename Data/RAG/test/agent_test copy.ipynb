```python
# API 키를 환경변수로 관리하기 위한 설정 파일
from dotenv import load_dotenv

# API 키 정보 로드
load_dotenv()
```

```python
from langchain.embeddings import OpenAIEmbeddings
from langchain.vectorstores import FAISS
from langchain.document_loaders import DirectoryLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain_openai import OpenAI

# ✅ 1. 여러 개의 문서 로드
loader = DirectoryLoader("documents/", glob="*.txt")  # 'documents' 폴더에서 모든 .txt 파일 로드
docs = loader.load()

# ✅ 2. 문서 분할 (너무 긴 문서는 잘라서 저장)
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
documents = text_splitter.split_documents(docs)

# ✅ 3. 문서 벡터화 (임베딩 생성)
embeddings = OpenAIEmbeddings()  # OpenAI API 키 필요
vector_db = FAISS.from_documents(documents, embeddings)

# ✅ 4. 사용자 질문에 대한 검색 수행
query = "AI 투자의 최신 트렌드는 무엇인가?"
retrieved_docs = vector_db.similarity_search(query, k=3)  # 관련 문서 3개 검색

# ✅ 5. 검색된 문서를 기반으로 GPT 모델을 사용하여 답변 생성
llm = OpenAI(model="gpt-4o")
context = "\n\n".join([doc.page_content for doc in retrieved_docs])  # 검색된 문서 결합
final_prompt = f"다음 문서를 참고하여 질문에 답변하세요:\n\n{context}\n\n질문: {query}\n답변:"
response = llm.invoke(final_prompt)

# ✅ 결과 출력
print("🔍 검색된 문서 기반 답변:")
print(response)
```
