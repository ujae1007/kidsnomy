```python
import os
import requests
from bs4 import BeautifulSoup

# 🔹 sources.txt에서 URL 리스트 불러오기
with open("sources.txt", "r") as f:
    urls = [line.strip() for line in f.readlines() if line.strip()]  # 빈 줄 제거

# 🔹 저장할 폴더 생성 (없으면 생성)
output_folder = "downloaded_texts"
os.makedirs(output_folder, exist_ok=True)

# 🔹 각 URL 처리
for idx, url in enumerate(urls):
    try:
        response = requests.get(url, timeout=10)  # 10초 타임아웃
        response.raise_for_status()  # HTTP 오류 발생 시 예외 처리

        # 🔹 HTML 파싱 (BeautifulSoup 활용)
        soup = BeautifulSoup(response.text, "html.parser")

        # 🔹 텍스트 추출 (웹페이지에서 주요 본문만 가져오기)
        paragraphs = soup.find_all("p")  # <p> 태그의 내용만 추출
        text_content = "\n".join([p.get_text() for p in paragraphs])

        # 🔹 저장할 파일명 지정
        filename = os.path.join(output_folder, f"document_{idx + 1}.txt")

        # 🔹 텍스트 저장
        with open(filename, "w", encoding="utf-8") as f:
            f.write(text_content)

        print(f"✅ {url} -> 저장 완료 ({filename})")

    except requests.exceptions.RequestException as e:
        print(f"❌ {url} 에서 데이터를 가져오는 중 오류 발생: {e}")
```
